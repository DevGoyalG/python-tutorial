## Pandas - DataFrame and Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame.
Series is a one-dimensional array-like object.
DataFrame is a two-dimensional, size-mutable and potentially heterogenous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

### Create a Series

```python
### Create a Series

import pandas as pd
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series: \n", series)
```

### Create a Series from Dictionary

```python
### Create a Series from Dictionary

data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ['a', 'b', 'c']
pd.Series(data, index)
```

### Create a DataFrame from dictonary of list

```python
### DataFrame

### Create a DataFrame from dictonary of list

data = {
    'Name': ['Dev', 'Ujjawal', 'Vaidik'],
    'Age': [22, 21, 20],
    'City': ['Mumbai', 'Bangalore', 'Pune']
}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

### Create a DataFrame from a list of Dictionaries

```python
### Create a DataFrame from a list of Dictionaries

data = [
    {'Name': 'Dev', 'Age': 22, 'City': 'Mumbai'},
    {'Name': 'Rishabh', 'Age': 21, 'City': 'Bangalore'},
    {'Name': 'Mreegendra', 'Age': 20, 'City': 'Kolkata'},
]
df = pd.DataFrame(data)
print(df)
print(type(df))
```

### Accessing data from DataFrame

```python
### Accessing data from DataFrame
data = {
    'Name': ['Dev', 'Ujjawal', 'Vaidik'],
    'Age': [22, 21, 20],
    'City': ['Mumbai', 'Bangalore', 'Pune']
}
df = pd.DataFrame(data)
print(df)
```

```python
df['Name']
```

```python
df.loc[0]          # row index
```

```python
df.loc[0, 'Name']
```

```python
df.iloc[0]       # column index
```

```python
df.iloc[0, 0]
```

```python
### Accessing a specified element using at

df.at[1, 'Age']
```

```python
### Accessing a specified element using iat

df.iat[2,2]
```
