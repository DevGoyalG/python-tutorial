# Reading Data from various data source using Pandas

## Reading JSON Data into a DataFrame

```python
import pandas as pd 
from io import StringIO
Data = '{"employee_name": "Rohit", "email": "rohitsharma45@gmail.com", "job_profile": [{"title1": "Captain", "title2": "Show Host"}]}'
df = pd.read_json(StringIO(Data))
```

```python
df
```

### Converting DataFrame to JSON

```python
df.to_json()
```

### Converting DataFrame to JSON Using Index Orientation

```python
df.to_json(orient='index')
```

### Converting DataFrame to JSON Using Records Orientation

```python
df.to_json(orient='records')
```

## Reading CSV Data from a URL

```python
df = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data", header=None)
```

```python
df
```

```python
df.head()
```

### Saving DataFrame to a CSV File

```python
df.to_csv("wine.csv")
```

```python
df
```

## Reading HTML Tables from Wikipedia

```python
url3 = "https://en.wikipedia.org/wiki/Mobile_country_code"

tables = pd.read_html(
    url3,
    storage_options={"User-Agent": "Mozilla/5.0"}
)

print(tables)
```

```python
df = tables[0]
df.head()
```

```python
len(tables)
```

```python
for i, table in enumerate(tables):
    print(f"Table {i}:")
    print(table.head())
```

### Extracting a Specific HTML Table Using Match and Header

```python
df = pd.read_html(
    url3,
    match="Country",
    header=0,
    storage_options={"User-Agent": "Mozilla/5.0"}
)[0]

df
```
