Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides wide range of functions for data manipulation and analysis, making it easier to clean, transform and extract insights from data.

### Handling missing values

```python
### Handling missing values

df[df.isnull()]         # True means missing values
```

```python
df.isnull().any(axis=1)
```

### Total missing values in each column

```python
# Total missing values in each column
df.isnull().sum()
```

### Fill all missing values

```python
# Fill all missing values
df_filled = df.fillna(0)
```

### Fill missing values with the mean of the column

```python
# Fill missing values with the mean of the column

df['Category_fillNA'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

### Renaming columns

```python
### Renaming columns

df = df.rename(columns= {'Date': 'Sales Date'})
df.head()
```

### Change data types

```python
### Change data types

df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
df['New Value'] = df['Value'].apply(lambda x:x*2)
df.head()
```

### Data Aggregating and Grouping

```python
### Data Aggregating and Grouping

df.head()
```

```python
grouped_mean = df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

### Grouping based on 2 categories - sum

```python
### Grouping based on 2 categories - sum

grouped_sum = df.groupby(['Product', 'Region'])['Value'].sum()
print(grouped_sum)
```

### Grouping based on 2 categories - mean

```python
### Grouping based on 2 categories - mean

grouped_mean2 = df.groupby(['Product', 'Region'])['Value'].mean()
print(grouped_mean2)
```

### Aggregate multiple functions

```python
### Aggregate multiple functions

grouped_agg = df.groupby('Region')['Value'].agg(['mean', 'sum', 'count'])
grouped_agg
```

### Merging and Joining DataFrames

```python
### Merging and Joining DataFrames

df1 = pd.DataFrame({'Key': ['A','B','C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'Key': ['A','B','D'], 'Value2': [4,5,6]})
```

```python
df1
```

```python
df2
```

### Merge DataFrame on the 'Key Columns'

```python
### Merge DataFrame on the 'Key Columns'

pd.merge(df1, df2, on="Key", how="inner")
```

```python
pd.merge(df1, df2, on="Key", how="outer")
```

```python
pd.merge(df1, df2, on="Key", how="left")
```

```python
pd.merge(df1, df2, on="Key", how="right")
```
