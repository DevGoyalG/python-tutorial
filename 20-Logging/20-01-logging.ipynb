## Python Logging

Logging is a crucial aspect of any application, providing a way to track events, errors and operational information. Python's built-in looging module offers a flexible framework  for emittng log messages from Python programs.

```python
import logging
```

### Configure the basic logging settings

```python
### Configure the basic logging settings

logging.basicConfig(level=logging.DEBUG)
```

### log messages with different severity levels

```python
### log messages with different severity levels

logging.debug("This is a debug message.")
logging.info("This is an info message.")
logging.warning("This is a warning message.")
logging.error("This is an error message.")
logging.critical("This is a critical message.")
```

### Log Levels

Python's logging module has several log levels indicating severity of events. The default levels are:
1. DEBUG    : Detailed info, typically of interest only when diagnosing problems.
2. INFO     : Confirmation that thngs are working as expected.
3. WARNING  : An indication that something unexpected happened or indicative of some problems in the near future (e.g., 'disk space low'). The software is still working as expected.
4. ERROR    : Due to a more serious problem, the software has not been able to pe
5. CRITICAL : A very serious error. indicating that the program itself may be unable to continue running.
