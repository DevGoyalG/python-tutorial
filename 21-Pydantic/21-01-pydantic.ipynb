# Pydantic Basics

Pydantic models are the foundation of data validation in Python. They use Python type annotations to define the structure and validate data at runtime.

## 1. Basic Pydantic Model: Creating and Using Models

### Creating Data Classes in Python

```python
### Creating Data Classes in Python

from dataclasses import dataclass

@dataclass
class Person():
    name:str
    age:int
    city:str

person = Person(name="Rohit Sharma", age=39, city="Mumbai")
print(person)
```

```python
person1 = Person(name="Virat Kohli", age=37, city=18)
print(person1)
```

### Creating and Validating Data with Pydantic

```python
from pydantic import BaseModel
```

```python
### Creating and Validating Data with Pydantic

class Person(BaseModel):
    name:str
    age:int
    city:str

person = Person(name="Rohit Sharma", age=39, city="Mumbai")
print(person)
```

```python
type(person)
```

```python
person1 = Person(name="Virat Kohli", age=37, city=18)
print(person1)
```

### Comparing Dataclass and Pydantic Validation


| Dataclass | Pydantic |
|---|---|
| Mainly used to create data containers | Used to create and validate data models |
| Does not enforce type hints at runtime | Validates data types at runtime |
| `city=18` is accepted even when `city: str` | `city=18` raises a validation error |
| Lightweight and simple | Provides data validation and parsing |

## 2. Model with Optional Fields

### Add Optional fields using Python's Optional type

```python
from typing import Optional

class Employee(BaseModel):
    id: int
    name: str
    dept: str
    salary: Optional[float] = None       # Optional with default value
    is_active: Optional[float] = True    # Optional with default value
```

### Creating a Model Without Optional Fields

```python
### Creating a Model Without Optional Fields

emp1 = Employee(id=1, name="Iron Man", dept="Stark Industries")
print(emp1)
```

### Creating a Model with Optional Fields

```python
### Creating a Model with Optional Fields

emp2 = Employee(id=2, name="Captain America", dept="Military", salary=60000.0, is_active=False)
print(emp2)
```

### Definition:
- Optional[Type]: Indicates the field can be None
- Default value (= None or = True): Makes the field optional
- Required fields must still be provided
- Pydantic validates types even for optional fields when values are provided

## 3. Model with List and Tuple Fields

### Defining a Pydantic Model with List Fields

```python
from pydantic import BaseModel
from typing import List

class Classroom(BaseModel):
    room_no: str
    students: List[str]         # List of strings
    capacity: int
```

### Working with List Fields

```python
### Working with List Fields
classroom = Classroom(room_no="B303", students=["Ujjawal", "Vaidik", "Mreegendra"], capacity=30)
print(classroom)
```

### Working with Tuple Fields

```python
### Working with Tuple Fields
classroom = Classroom(room_no="B303", students=("Ujjawal", "Vaidik", "Mreegendra"), capacity=30)
print(classroom)
```

### Validating List Field Values

```python
### Validating List Field Values
try:
    invalid_val = Classroom(room_no="B1", students=["Dev", 223], capacity=30)
except ValueError as e:
    print(e)
```

## 4. Model with Nested Models

### Create complex structures with Nested Models

```python
### Create complex structures with Nested Models
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    zip_code: str

class Customer(BaseModel):
    customer_id: int
    name: str
    address: Address            # Nested model
```

### Create a customer with nested address

```python
# Create a customer with nested address
customer = Customer(
    customer_id=1, 
    name="Thor", 
    address={"street": "1 Main St", "city": "Asgard", "zip_code": "001"}
)
print(customer)
```

## 5. Pydantic Fields: Customization and Constraints

The field function in the Pydantic enhances model fields beyond baisc type hints by allowing you to specify validation rules, default values, alliases and more.

### Defining Field Constraints

```python
from pydantic import BaseModel, Field

class Item(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    price: float = Field(gt=0, le=1000)                  # greater than 0, less than or equal to 1000
    quantity: int = Field(gt=0)                          # greater than 0

# Valid instance
item = Item(name="Book", price=45.18, quantity=7)
print(item)
```

### Customizing Fields with Defaults and Descriptions

```python
from pydantic import BaseModel, Field

class User(BaseModel):
    username: str = Field(..., description="Unique username for the user")
    age: int = Field(default=18, description="User age, defaults to 18")
    email: str = Field(default_factory=lambda: "user@example.com", description="Default email address")
```

### Creating a Model with Default Values

```python
### Creating a Model with Default Values
user1 = User(username="iron_man")
print(user1)
```

### Creating a Model with All Fields

```python
### Creating a Model with All Fields
user2 = User(username="captain_america", age=70, email="captain@gmail.com")
print(user2)
```

### Generating a JSON Schema

```python
### Generating a JSON Schema
print(User.model_json_schema())
```
